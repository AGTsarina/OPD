### Метод Якоби

```python
import numpy as np

def jacobi(A, b, x0=None, eps=1e-8, max_iter=1000):
    A = np.asarray(A, dtype=float)
    b = np.asarray(b, dtype=float)
    n = len(b)
    x = np.zeros(n) if x0 is None else np.array(x0, dtype=float)
    for _ in range(max_iter):
        x_new = np.zeros(n)
        for i in range(n):
            s = (A[i, :i] @ x[:i]).sum() + (A[i, i+1:] @ x[i+1:]).sum()
            x_new[i] = (b[i] - s) / A[i, i]
        if np.linalg.norm(x_new - x, np.inf) < eps:
            return x_new
        x = x_new
    return x
```

```python
A = np.array([[4, -1,  0],
              [-1, 4, -1],
              [0, -1,  4]], dtype=float)
b = np.array([2, 4, 10], dtype=float)

x = jacobi(A, b)
A @ x
```

### Метод Зейделя

```python
def seidel(A, b, x0=None, eps=1e-8, max_iter=1000):
    A = np.asarray(A, dtype=float)
    b = np.asarray(b, dtype=float)
    n = len(b)
    x = np.zeros(n) if x0 is None else np.array(x0, dtype=float)
    for _ in range(max_iter):
        x_old = x.copy()
        for i in range(n):
            s = (A[i, :i] @ x[:i]).sum() + (A[i, i+1:] @ x[i+1:]).sum()
            x[i] = (b[i] - s) / A[i, i]
        if np.linalg.norm(x - x_old, np.inf) < eps:
            return x
    return x
```

```python
x = seidel(A, b)
A @ x
```

```python
def sor(A, b, omega=1.5, x0=None, eps=1e-8, max_iter=1000):
    A = np.asarray(A, dtype=float)
    b = np.asarray(b, dtype=float)
    n = len(b)
    x = np.zeros(n) if x0 is None else np.array(x0, dtype=float)
    for _ in range(max_iter):
        x_old = x.copy()
        for i in range(n):
            s = (A[i, :i] @ x[:i]).sum() + (A[i, i+1:] @ x[i+1:]).sum()
            x_gs = (b[i] - s) / A[i, i]
            x[i] = (1 - omega) * x[i] + omega * x_gs
        if np.linalg.norm(x - x_old, np.inf) < eps:
            return x
    return x
```

```python
x = sor(A, b)
A @ x
```
